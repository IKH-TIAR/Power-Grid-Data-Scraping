```python
import requests
from bs4 import BeautifulSoup
import csv
import urllib3
import time

# Disable SSL warnings (temporary solution)
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

# Configure settings
base_url = "https://erp.powergrid.gov.bd/web/generations/view_generations"
total_pages = 1776
output_file = "all_pages_data.csv"
delay_between_requests = 1  # seconds to wait between requests

# Create session for connection pooling
session = requests.Session()

# Set common headers
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
}

def scrape_page(page_number):
    """Scrape data from a single page"""
    try:
        # Add pagination parameter (adjust if needed - check website's actual pagination pattern)
        params = {'page': page_number}
        
        response = session.get(
            base_url,
            params=params,
            headers=headers,
            verify=False,
            timeout=30
        )
        response.raise_for_status()
        
        soup = BeautifulSoup(response.content, 'html.parser')
        table = soup.find('table')
        
        if not table:
            print(f"No table found on page {page_number}")
            return None
            
        # Extract data
        page_data = []
        for row in table.find_all('tr')[2:]:  # Skip header row
            cols = row.find_all(['td', 'th'])[:4]
            page_data.append([col.get_text(strip=True) for col in cols])
            
        return page_data
        
    except Exception as e:
        print(f"Error on page {page_number}: {str(e)}")
        return None

# Initialize CSV with headers
with open(output_file, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    
    # Get headers from first page
    first_page_data = scrape_page(1)
    if first_page_data:
        header_row = [h.get_text(strip=True) for h in table.find('tr').find_all(['th', 'td'])[:4]]
        writer.writerow(header_row)
        writer.writerows(first_page_data)

# Scrape remaining pages
for page in range(2, total_pages + 1):
    try:
        page_data = scrape_page(page)
        if page_data:
            with open(output_file, 'a', newline='', encoding='utf-8') as csvfile:
                writer = csv.writer(csvfile)
                writer.writerows(page_data)
            print(f"Page {page}/{total_pages} scraped successfully")
            
        # Be polite - add delay
        time.sleep(delay_between_requests)
        
    except KeyboardInterrupt:
        print("Scraping interrupted by user")
        break
    except Exception as e:
        print(f"Critical error on page {page}: {str(e)}")
        continue

print(f"Scraping completed. Data saved to {output_file}")
```
